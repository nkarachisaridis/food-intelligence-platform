# Explore data

## 1 Connect DuckDB to the file

```python
import duckdb
con = duckdb.connect()
PATH = "/workspace/data/food.parquet"
```

```python
con.sql(
    f" select count(*) as products from '{PATH}'").df()
```

```python
con.sql(f" SELECT * FROM '{PATH}'" ).df()
```

```python
con.sql(f"DESCRIBE SELECT * FROM '{PATH}'").df()
```

```python
con.sql(f"SELECT * FROM '{PATH}' LIMIT 5").df().T
```
